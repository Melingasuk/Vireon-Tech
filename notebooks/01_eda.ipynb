# Exploratory Data Analysis: Soil Health & Crop Recommendation

This notebook explores the soil health dataset and identifies patterns in nutrient levels (NPK) and their relationship with suitable crops.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## 1. Load Data

```python
# Load the soil data
df = pd.read_csv('../data/raw/soil_data.csv')

# Display basic info
print(f"Dataset shape: {df.shape}")
print(f"\nFirst 5 rows:")
print(df.head())
print(f"\nData types:")
print(df.dtypes)
print(f"\nMissing values:")
print(df.isnull().sum())
```

## 2. Statistical Summary

```python
# Statistical summary
print(df.describe())
```

## 3. Nutrient (NPK) Distribution

```python
# Visualize NPK distribution
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for idx, nutrient in enumerate(['N', 'P', 'K']):
    axes[idx].hist(df[nutrient], bins=30, edgecolor='black', alpha=0.7)
    axes[idx].set_title(f'{nutrient} Distribution')
    axes[idx].set_xlabel(f'{nutrient} Level')
    axes[idx].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

## 4. Crop Distribution

```python
# Count crops
crop_counts = df['Crop_Type'].value_counts()
print(crop_counts)

# Visualize
plt.figure(figsize=(10, 6))
crop_counts.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Crop Distribution in Dataset')
plt.xlabel('Crop Type')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 5. NPK vs Crop Type

```python
# Box plots for NPK by crop
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for idx, nutrient in enumerate(['N', 'P', 'K']):
    df.boxplot(column=nutrient, by='Crop_Type', ax=axes[idx])
    axes[idx].set_title(f'{nutrient} by Crop Type')
    axes[idx].set_xlabel('Crop Type')
    axes[idx].set_ylabel(f'{nutrient} Level')

plt.suptitle('')
plt.tight_layout()
plt.show()
```

## 6. Correlation Analysis

```python
# Correlation matrix
numeric_cols = ['N', 'P', 'K', 'pH', 'Rainfall', 'Temperature', 'Humidity']
correlation = df[numeric_cols].corr()

# Visualize
plt.figure(figsize=(10, 8))
sns.heatmap(correlation, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Feature Correlation Matrix')
plt.tight_layout()
plt.show()
```
